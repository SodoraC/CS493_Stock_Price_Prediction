```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import yfinance as yf
```

Below is an example of how to use the Yfinance ticker method. The info method tells us all the features that we can access with the get info method.

```python
Verizon = yf.Ticker("VZ")

print(Verizon.info)

```

```python
stocks = ["VZ"]

price_movement_columns = ["Open", "High", "Low", "Close", "Volume"]

summary_metrics = [
    "dividendYield", "payoutRatio", "fiveYearAvgDividendYield", "trailingPE",
    "forwardPE", "marketCap", "fiftyTwoWeekLow", "fiftyTwoWeekHigh",
    "profitMargins", "shortRatio", "priceToBook", "earningsQuarterlyGrowth",
    "trailingEps", "forwardEps", "targetHighPrice", "targetLowPrice",
    "targetMeanPrice", "recommendationKey", "totalCashPerShare", "totalDebt",
    "earningsGrowth", "revenueGrowth", "fullTimeEmployees"
]

for ticker in stocks:

    price_movement_df_full = yf.download(ticker, start="2015-11-05", end="2025-11-05")
    historical_df_selected = price_movement_df_full[price_movement_columns]

    hist_filename = f"{ticker}_PRICE_MOVEMENT_Data.csv"
    historical_df_selected.to_csv(hist_filename)


    stock = yf.Ticker(ticker)


    quarterly_fin = stock.quarterly_financials
    fin_filename = f"{ticker}_QUARTERLY_FINANCIALS.csv"
    quarterly_fin.T.to_csv(fin_filename)

    quarterly_bs = stock.quarterly_balance_sheet
    bs_filename = f"{ticker}_QUARTERLY_BALANCE_SHEET.csv"
    quarterly_bs.T.to_csv(bs_filename)


    info = stock.info

    summary_data = {}
    for metric in summary_metrics:
        summary_data[metric] = [info.get(metric, 'N/A')]

    summary_df = pd.DataFrame(summary_data)

    summary_filename = f"{ticker}_SUMMARY_METRICS.csv"
    summary_df.to_csv(summary_filename, index=False)
```
