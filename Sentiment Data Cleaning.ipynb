```python
import pandas as pd
```

```python
df = pd.read_csv("VZ Sentiment Data.csv")
df['date'] = pd.to_datetime(df['date']).dt.date
df = df.sort_values(by=['date'])
df.reset_index(drop=True, inplace=True)
df
```

```python
df_removed_pipes = df
df_removed_pipes['title'] = df['title'].str.split('|').str[0]
df_duplicates = df[df_removed_pipes.duplicated()]
df_removed_duplicates = df_removed_pipes.drop_duplicates()
print(df_duplicates)
df_removed_duplicates
```

```python
df_removed_duplicates.to_csv("VZ Sentiment Data Removed Duplicates.csv", index=False)
```
